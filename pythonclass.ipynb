```python
students = {
    101: ("mukesh", [67, 79, 49, 58, 77]),
    102: ("raja", [87, 77, 45, 58, 67]),
    103: ("roshan", [67, 66, 85, 56, 47]),
    104: ("vivek", [67, 76, 95, 76, 77]),
    105: ("rahul", [60, 76, 49, 51, 57])
}
# function to add a new student.
def add_student():
    name = input("Enter student name: ")  # ask user for student name.
    roll = int(input("Enter roll number: ")) # ask for roll number.

    if roll in students:                #check if roll number already exists.
        print("Student already exists.") #if yes show msg
        return    #stop this function

    grades = input("Enter grades (comma-separated): ") #ask user to enter marks.
    grades_list = [int(g.strip()) for g in grades.split()] #convert string to list of number.

    students[roll] = (name, grades_list) #store name and grades in the distionary.
    print("Student added.") # show success msg.

def list_student(): #function to list all students.
    if not students: #check if no student is present.
        print("No students found.")   #show msg if empty.
        return  #exit the function.
    for roll, (name, grades) in students.items(): # loop through each student and print their info.
        print(f"Roll Number: {roll}, Name: {name}, Grades: {grades}") #ptint student details.
#function to show avg marks of one student.
def average_grade():
    roll = int(input("Enter roll number: ")) # ask for roll number.
    if roll not in students: #check if roll number not found.
        print("Student not found.") # show msg
        return # exit.
    name, grades = students[roll] # get the name and marks of the student.

    average = sum(grades) / len(grades) # calculate avg = total marks divide number of subject.
    print(f"Average for {name}: {average:.2f}") # print avg marks .
# function to find the student with the highest avg marks .
def top_student():
    if not students: #if no student is added yet.
        print("No students found.") #show msg.
        return # exit.
#find the roll number of the student with highest avg.
    top_roll = max(students, key=lambda r: sum(students[r][1]) / len(students[r][1]))
    name, grades = students[top_roll] # get name and marks of top student.
    avg = sum(grades) / len(grades)  # calculate avg.
    print(f"Top student: {name} (Roll: {top_roll}), Average: {avg:.2f}") # print result.
#function to show the manu and take user input.
def show_menu():
    while True:     # run this loop again and again
        print("\nMenu:")
        print("1. Add Student")
        print("2. List Students")
        print("3. Average Grade")
        print("4. Top Student")
        print("5. Exit")

        choice = input("Enter your choice: ") # take user choice.

        if choice == "1":
            add_student()    # call function to add student.
        elif choice == "2":
            list_student()   # call function to list all student.
        elif choice == "3":
            average_grade()  # call function to show one student avg.
        elif choice == "4":
            top_student()    # call function to show top student.
        elif choice == "5":
            print("Exiting.") # exit msg.
            break  # exit the loop and program.
        else:
            print("Invalid choice.") # if wrong option is entered.

# start the program bt showing the manu.
show_menu()
```

```python
class Car:
    # Constructor to initialize the object
    def __init__(self, brand1, model1):
        print('Inside init...')
        self.brand = brand1
        self.model = model1


    # Method to display car details
    def display(self):
        print(self.model)
        print(self.brand)


c = Car('TATA', 2025)
```
